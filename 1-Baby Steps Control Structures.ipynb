```python
BMI=int(input("Enter the BMI Index:"))
if(BMI<25):
    print("Overweight")
elif(BMI<35):
    print("Very Overweight")
elif(BMI<18.5):
    print("underweight")
elif(BMI<22.9):
    print("Normal")
else:
    print("Obese")
```
