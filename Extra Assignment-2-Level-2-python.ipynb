```python
# print Odd Numbers in the list
```

```python
L=(20, 10, 16, 19, 25, 1, 276, 188)
print(L)
for n in L:
    if ((n%2)!=0):
        print(n,"is Odd")

```

```python
# print Even numbers in the list
```

```python
L=(20, 10, 16, 19, 25, 1, 276, 188)
print(L)
for n in L:
    if ((n%2)==0):
        print(n,"is Even")
```
